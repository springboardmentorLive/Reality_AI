•	Store this dataset into a CSV file named raw_data.csv.
•	Read the first 5 rows of the data from raw_data.csv using pandas.
•	Drop rows with any missing values.
•	Count the number of missing values in each column.
•	Fill missing values in the Name column with "some name" and in the Age column with 21.
•	Save the cleaned dataset into a new CSV file named cleaned_data.csv.

```python
import pandas as pd

data = {
    'Name': ['Riya', 'Karan', None, 'Meena', 'Amit', None, 'Sneha', 'Vikas', 'Neha', 'Tina'],
    'Age': [28, 30, 22, None, 26, 21, None, 24, 29, None],
    'Gender': ['F', 'M', 'M', 'F', 'M', 'M', 'F', 'M', 'F', 'F'],
    'Score': [91, None, 76, 84, 89, 95, None, 88, 93, 90]
}

df = pd.DataFrame(data)
df.to_csv('raw_data.csv', index=False)
```

```python
df.dropna()
```

```python
df.isnull().sum()
```

```python
cleaned_df = df.copy()

cleaned_df["Name"].fillna("Anshu", inplace=True)
cleaned_df["Age"].fillna(21,inplace=True)
cleaned_df
```

```python
cleaned_df.to_csv("cleaned_data.csv")
```
