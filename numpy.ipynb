```python
#matplotlib
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import style

ipl_teams={'SRH':289,'MI':266,'KKR':280,'CSK':260}
runs_score=list(ipl_teams.values())
teams=list(ipl_teams.keys())
plt.bar(teams,runs_score,color='green')#,width=we give what we want
```

```python
style.use('ggplot')
plt.figure(figsize=(15,10))
plt.xlabel('team')
plt.ylabel('runs')
plt.bar(teams,runs_score,color='green')
```

```python
style.use('ggplot')
plt.figure(figsize=(15,10))
plt.xlabel('team')
plt.ylabel('runs')
plt.bar(teams,runs_score,color='green',width=0.2)
```

```python

style.use('ggplot')
plt.xlabel('team')
plt.ylabel('runs')
c=['orange','blue','darkblue','yellow']
l=['SRH','MI','KKR','CSK']
plt.bar(teams,runs_score,color=c,width=0.2,label=l)
plt.legend()
```

```python
import sys
# Ensure `matplotlib.pyplot` module is completely reloaded to fix potential corruption.
if 'matplotlib.pyplot' in sys.modules:
    del sys.modules['matplotlib.pyplot']
if 'plt' in globals():
    del plt

import matplotlib.pyplot as plt
import csv
Team1=[]
Team1_runs=[]

with open('/content/scores.csv','r') as csvfile:
    plots=csv.reader(csvfile,delimiter=',')
    next(plots)
    for row in plots:
      try:
        Team1.append(row[1])
        Team1_runs.append(int(row[2]))
      except:
        pass


plt.pie(Team1_runs,labels=Team1,autopct='%2f%%')
plt.title('IPL 2020')
plt.axis('equal')
plt.show()
```
