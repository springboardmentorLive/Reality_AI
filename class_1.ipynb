```python
import pandas as pd
left = pd.DataFrame(
    {
        "key":["K0", "K1", "K2", "AЗ"],
        "Α": ["Α0", "Α1", "A2", "A3"],
        "B": ["B0", "B1", "B2", "B3"],
    }
)

right = pd.DataFrame(
    {
       "key": ["K0", "К1", "K2", "К3"],
       "C": ["C0", "C1", "C2", "C3"],
       "D": ["D0", "D1", "D2", "D3"],
    }
)

result = pd.merge(left, right, on="key")

result
```

```python
left=pd.DataFrame(
{
"key": ["K0", "K1", "K2", "A3"],
"A": ["A0", "A1", "A2", "A3"],
"B":["B0", "B1", "B2", "B3"],
}
)

right=pd.DataFrame(
{
"key": ["K0", "K1", "K2", "K3"],
"C": ["C0", "C1", "C2", "C3"],
"D": ["D0", "D1", "D2", "D3"],
}
)

result = pd.merge(left, right, on="key")

result
```

```python
left=pd.DataFrame(
{
"key1": ["K0", "K0", "K1", "K2"],
"key2": ["K0", "K1", "K0", "К1"],
"A": ["A0", "A1", "A2", "A3"],
"B":["B0", "B1", "B2", "B3"],
}
)

right=pd.DataFrame(
{
"key1": ["K0", "K1", "K1", "K2"],
"key2": ["K0", "K0", "K0", "K0"],
"C": ["C0", "C1", "C2", "C3"],
"D": ["D0", "D1", "D2", "D3"],
}
)

result = pd.merge(left, right, how="left", on=["key1", "key2"])

result
```

```python
result = pd.merge(left, right, how="right", on=["key1", "key2"])
result
```

```python
result = pd.merge(left, right, how="outer", on=["key1", "key2"])
result
```

```python
result = pd.merge(left, right, how="inner", on=["key1", "key2"])
result
```
