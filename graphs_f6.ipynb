```python
import matplotlib.pyplot as pd
colors = ["Red", "Blue", "Green", "Yellow"]
counts = [12, 25, 18, 7]
bar_colors = [
    (1, 0, 0),    # red
    (0, 0, 1),    # blue
    (0, 1, 0),    # green
    (1, 1, 0)     # yellow
]
```

```python
plt.style.use('ggplot')
plt.figure(figsize=(5,5))
plt.bar(colors, counts, color=bar_colors)
plt.xlabel('colors')
plt.ylabel('count')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

days = [0, 1, 2, 3, 4]

FB = [350, 80, 360, 190, 70]
YT = [360, 90, 330, 120, 60]
LG = [80, 95, 60, 90, 40]
SC = [70, 80, 250, 85, 30]

x = np.arange(len(days))
w = 0.2
```

```python
plt.bar(x, FB, width=w, label="FB")
plt.bar(x + w, YT, width=w, label="YT")
plt.bar(x + 2*w, LG, width=w, label="LG")
plt.bar(x + 3*w, SC, width=w, label="SC")

plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import csv
import requests
from io import StringIO

url = "https://raw.githubusercontent.com/selva86/datasets/master/Scores.csv"
response = requests.get(url)

Subjects = []
Scores = []

f = StringIO(response.text)
lines = csv.reader(f, delimiter=',')
next(lines)
for row in lines:
    Subjects.append(row[0])
    Scores.append(int(row[1]))

plt.figure(figsize=(8,5))
plt.bar(Subjects, Scores, color='skyblue')
plt.title('Marks of a Student', fontsize=20)
plt.xlabel('Subjects')
plt.ylabel('Scores')
plt.show()
```
