```python
import pandas as pd
left = pd.DataFrame(
    {
        "key": ["K0","K1", "K2", "K3"],
        "A": ["A0", "A1", "A2", "A3"],
        "B": ["B0", "B1", "B2", "B3"],
    }
)

right = pd.DataFrame(
    {
        "key": ["K0", "K1", "K2", "K3"],
        "C": ["C1", "C2", "C3", "C4"],
        "D": ["D1", "D2", "D3", "D4"],
    }
)

result = pd.merge(left, right, on="key")
result
```

```python
import pandas as pd
left = pd.DataFrame(
    {
        "key1": ["K0","K1", "K2", "K3"],
        "key2": ["K0", "K1", "K2", "K3"],
        "A": ["A0", "A1", "A2", "A3"],
        "B": ["B0", "B1", "B2", "B3"],
    }
)

right = pd.DataFrame(
    {
        "key1": ["K0", "K1", "K2", "K3"],
        "key2": ["K0", "K1", "K2", "K3"],
        "C": ["C1", "C2", "C3", "C4"],
        "D": ["D1", "D2", "D3", "D4"],
    }
)

result = pd.merge(left, right, how="left", on=["key1", "key2"])
result
```

```python
result = pd.merge(left, right, how="right", on=["key1", "key2"])
result
```

```python
result = pd.merge(left, right, how="inner", on=["key1", "key2"])
result
```

```python
result = pd.merge(left, right, how="outer", on=["key1", "key2"])
result
```

```python
import pandas as pd
data = {'Car_Name': ['Tata', 'Mahindra', 'BYD', 'Tesla'],
        'Mileage(km)': [100, None, 200, None],
        'Country': ['Delhi', None, 'China', 'America'],
        'Fuel_Type': ['Petrol', 'Petrol', 'EV', 'EV']}
print(data)
```

```python
df = pd.DataFrame(data)
print(df)
```

Step-by-Step Data Cleaning Method
1. Check for Missing Data:
Shows which cells are missing(True)


```python
df.isna()  #it is used to check the null values in the dataframe
```

```python
df.isna().sum() #It is like an extension of isna() where we get the sum
```

2. Drop Missing Data

```python

df.dropna()  #it is used to drop the rows with null values and the output only shows the rows with zero null values
```

Drop Columns with Nan/None or Null values

```python
df.dropna(axis=1)
```

Drop Rows with Null values (default)

```python
df.dropna(axis = 0)
```

3. Drop Rows with missing values

```python
df.dropna(subset=['Country'])
```

```python
df.at[1,'Mileage(km)']=150 #at method is used to update a particuar cell
df                            #at[index_no., Col_name] = new value
```

4. Fill Missing values

```python
df.fillna('Unknown')  #fillna is used to fill values wherever null values are present
```

Rename Columns

```python
df.rename(columns={'Car_Name': 'Car_Brand'}, inplace=True)
df
```

```python
df['Car_Brand'] =df['Car_Brand'].str.upper()
df
```

```python
df['Car_Brand'] = df['Car_Brand'].str.lower()
df
```

```python
df = df[df['Mileage(km)'] < 200]
```
