```python
print("Chatbot: Hi! I am your chatbot. Type 'bye' to exit.")
name = input("Chatbot: Before we start, what is your name?\nYou: ")

print(f"Chatbot: Nice to meet you, {name}! How can I help you today?")

responses = {
    ("hi", "hello", "hey", "hii", "good morning", "good evening"):
        f"Chatbot: Hi {name}! How can I help you?",

    ("what is your name", "who are you", "your name"):
        "Chatbot: I am a simple chatbot created to help you.",

    ("who created you", "who made you", "your creator"):
        "Chatbot: I was created by a Python programmer like you!",

    ("how old are you", "your age", "what is your age"):
        "Chatbot: I don't have an age, I live inside your computer!",

    ("what is the time", "current time", "tell me time"):
        "Chatbot: I cannot tell time yet, but you can check your device clock.",

    ("what are your hobbies", "your hobbies"):
        "Chatbot: I like chatting with you!",

    ("your favorite color", "favourite colour", "what is your favorite color"):
        "Chatbot: I love all colors equally because I live in a digital world!",

    ("tell me a joke", "joke", "make me laugh"):
        "Chatbot: Why don’t computers take naps? Because they have too many tabs open!",

    ("what is ai", "explain ai", "meaning of ai"):
        "Chatbot: AI means making computers think and learn like humans.",

    ("help", "i need help", "can you help me"):
        f"Chatbot: Sure {name}! Tell me what you need help with.",

    ("how are you", "how r u", "how you doing"):
        f"Chatbot: I am doing great, {name}! Thanks for asking.",

    ("thank you", "thanks", "thx"):
        f"Chatbot: You're welcome, {name}!"
}

while True:
    user = input("You: ").lower().strip()

    if user in ["bye", "exit", "quit"]:
        print(f"Chatbot: Goodbye {name}! Have a great day.")
        break

    found = False
    for keys, reply in responses.items():
        if user in keys:
            print(reply)
            found = True
            break

    if not found:
        print(f"Chatbot: Sorry {name}, I didn't understand that. Try asking something else.")
```
