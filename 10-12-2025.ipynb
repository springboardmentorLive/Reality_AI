```python
print("Chatbot: Hi! I am your chatbot. Type 'bye' to exit.")
while True:
    user = input("You: ").lower()
    if user in ["hi", "hello", "hey", "hola", "namaste"]:
        print("Chatbot: Hello! How can I help you?")


    elif user in ["how are you", "how r u", "how's it going"]:
        print("Chatbot: I am fine! Thank you")


    elif user in ["bye", "goodbye", "see you", "exit"]:
        print("Chatbot: Goodbye! Have a nice day!")
        break


    elif user in ["what is your name", "who are you"]:
        print("Chatbot: I am a Robo 2.O created by Dhruv!")

    elif user in ["what can you do", "help"]:
        print("Chatbot: I can answer simple questions like greetings, my name, and more!")

    elif user in ["time", "current time"]:
        from datetime import datetime
        print("Chatbot: Current time is", datetime.now().strftime("%H:%M:%S"))

    else:
        print("Chatbot: Sorry, I don't understand.")
```
