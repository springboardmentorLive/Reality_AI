```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import style
```

```python
time_data={'facebook':120, 'youtube':90, 'instagram':40, 'snapchat':28}

social_media=list(time_data.keys())

time_speed=list(time_data.values())
```

```python
plt.bar(social_media,time_speed,color='maroon')
```

```python
style.use('ggplot')

plt.figure(figsize=(5,5))

plt.bar(social_media,time_speed, color='orange', width=0.25)

plt.xlabel('social media')

plt.ylabel('time')

plt.title('average time speed on social media')

Text=('0.5, 1.0, average time speed on social media')
```

```python
style.use('ggplot')

plt.figure(figsize=(5,5))

colors=['red','green','blue','yellow']

plt.bar(social_media, time_speed, color=colors, width=0.25)

plt.xlabel('social media')

plt.ylabel('time')

plt.title('average time speed on social media')

Text=('0.5, 1.0, average time speed on social media')
```

```python
import matplotlib.pyplot as plt
import csv

Author = []
Genre = []

with open('/content/book.csv', 'r') as csvfile:
    lines = csv.reader(csvfile, delimiter=',')
    for row in lines:
        Author.append(row[0])
        Genre.append(row[1])
plt.bar(Author, Genre, color=['blue', 'green', 'red', 'purple'])
plt.title(' Book ', fontsize=20)
plt.xlabel('Author')
plt.ylabel('Genre')
plt.show()
```

```python
import matplotlib.pyplot as plt
labels = ['pizza', 'biryani', 'icecream', 'choclate', 'burger']
sizes = [18.00, 19.00, 20.00, 30.20, 20.00]
plt.axis('equal')
plt.pie(sizes, labels=labels, autopct='%1.2f%%', startangle=90)
plt.title('Marks of a Student')
plt.show()
```
