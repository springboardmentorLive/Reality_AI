```python
import cv2
import os

# Path to input video file
video_path = "input.mp4"

# Folder where extracted images will be saved
output_folder = "frames_every_2s"
os.makedirs(output_folder, exist_ok=True)  # Create folder if doesn't exist

# Open the video file using OpenCV
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print(f"Error: Could not open video")
    exit()

# Get the video's Frames Per Second (FPS)
fps = cap.get(cv2.CAP_PROP_FPS)

# Choose how often to save a frame (in seconds)
interval_seconds = 2

# Convert seconds → number of frames to skip
interval_frames = int(fps * interval_seconds)

frame_num = 0  # Keeps track of current frame index
saved = 0      # Counts how many frames have been saved

while True:
    # Read one frame from the video
    success, frame = cap.read()

    # If no more frames are available, stop the loop
    if not success:
        break

    # Save only if the frame number matches the interval
    # Example: save frames 0, 60, 120, 180... (for 2s interval)
    if frame_num % interval_frames == 0:
        # Save the frame as an image file
        cv2.imwrite(f"{output_folder}/frame_{saved:05d}.jpg", frame)
        saved += 1  # Increase save counter

    frame_num += 1  # Move to next frame

# Release (close) the video file
cap.release()

print("Saved:", saved, "frames")
```
