```python
import cv2
import os
```

```python
from IPython.core.error import TryNext
video_path = "/content/AIvideo.mp4"
output_folder = 'frames'
os.makedirs(output_folder, exist_ok=True)
cap = cv2.VideoCapture(video_path)
if not cap.isOpened():
    print("Error: Could not open video.")
    exit()
frame_count = 0
while True:
  ret, frame = cap.read()
  if not ret:
    break # End of video
  # Save frame as image
  frame_filename = os.path.join(output_folder, f'frame_{frame_count:03d}.jpg')
  cv2.imwrite(frame_filename, frame)
  print(f"Saved {frame_filename}")
  frame_count += 1

# Release resources
cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import os

video_path = "/content/AIvideo.mp4"
save_dir = "/content/output_frames"
os.makedirs(save_dir, exist_ok=True)

cap = cv2.VideoCapture(video_path)
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

# Pick 5 evenly spaced positions
positions = [int(i * (total_frames - 1) / 4) for i in range(5)]

for i, pos in enumerate(positions):
    cap.set(cv2.CAP_PROP_POS_FRAMES, pos)
    ret, frame = cap.read()
    if ret:
        cv2.imwrite(f"{save_dir}/frame_{i+1}.jpg", frame)
        print(f"Saved frame_{i+1}.jpg in {save_dir}")

cap.release()
print("Done! 5 frames saved.")
```

```python
import cv2
from PIL import Image
from IPython.display import display

video_path = "/content/AIvideo.mp4"

cap = cv2.VideoCapture(video_path)
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

# 5 evenly spaced frame positions
positions = [int(i * (total_frames - 1) / 4) for i in range(5)]

for i, pos in enumerate(positions):
    cap.set(cv2.CAP_PROP_POS_FRAMES, pos)  # move to frame
    ret, frame = cap.read()

    if ret:
        print(f"Frame {i+1}:")
        img = Image.fromarray(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))
        display(img)  # EXACT same style output as before

cap.release()
```
