```python
import matplotlib.pyplot as plt
# Data from the image
labels = ['CSE', 'ISE', 'EEE', 'ECE', 'MATHS']
sizes = [18.07, 9.24, 22.49, 30.12, 20.08]
colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99', '#c2c2f0'] # Example colors
# Create the pie chart
plt.figure(figsize=(8, 6)) # Optional: adjust figure size
plt.pie(sizes, labels=labels, autopct='%1.2f%%', colors=colors)

# Add a title
plt.title('Marks of a Student')
# Ensure the chart is a circle
plt.axis('equal')
# Display the chart
plt.show()
```

```python
import cv2
import os

video_path="/content/Superhero_Video_Creation_Request.mp4"

output_folder="frames"

os.makedirs(output_folder,exist_ok=True)

cap=cv2.VideoCapture("/content/Superhero_Video_Creation_Request.mp4")

if not cap.isOpened():
  print("error opening video file")
  exit()
frame_count=0
while True:
  ret,frame=cap.read()
  if not ret:
    break
  frame_filename = os.path.join(output_folder, f"frame_{frame_count:04d}.jpg")
  cv2.imwrite(frame_filename, frame)
  print(f"saved frame {frame_count}")
  frame_count+=1

  if frame_count>=5:
    break

cap.release()
cv2.destroyAllWindows()
```
