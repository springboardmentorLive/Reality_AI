```python
import cv2
import os
import matplotlib.pyplot as plt
from IPython.display import display, Image
video_path="/content/Hailuo_Video_A tiny fluffy baby panda wavin_454426196678393863 (1).mp4"
output_folder = "frames1"
os.makedirs(output_folder, exist_ok=True)
cap = cv2.VideoCapture(video_path)
max_frames = 5
saved_frame_paths = []

if not cap.isOpened():
    print("Error: Could not open video. Please ensure the video path is correct and the file exists.")
else:
    frame_count = 0
    print(f"Extracting up to {max_frames} frames...")

    while frame_count < max_frames:
        ret, frame1 = cap.read()
        if not ret:
            break
        frame_filename = os.path.join(output_folder, f"frame1_{frame_count:03d}.jpg")
        cv2.imwrite(frame_filename, frame1)
        saved_frame_paths.append(frame_filename)
        print(f"Saved {frame_filename}")
        frame_count += 1
    cap.release()
    cv2.destroyAllWindows()
    print("\nExtraction complete. Displaying saved frames:")

if saved_frame_paths:
    for path in saved_frame_paths:
        print(f"--- {os.path.basename(path)} ---")
        display(Image(filename=path, width=300))
else:
    print("No frames were saved to display.")
```

```python
import cv2  # Imports OpenCV for image and video processing.
import os
import sys

# Path to your video file
video_path = "/content/Hailuo_Video_A tiny fluffy baby panda wavin_454426196678393863 (1).mp4"
# Folder to save extracted frames
output_folder = "frames"

# Create the folder if it doesn't exist
os.makedirs(output_folder, exist_ok=True)
print(f"Output folder '{output_folder}' ensured to exist.")

# Load the video
cap = cv2.VideoCapture(video_path)

# Check if video opened successfully
if not cap.isOpened():
    print(f"Error: Could not open video at {video_path}")
    sys.exit()

frame_count = 0
print("Starting frame extraction...")

while True:
    ret, frame = cap.read()

    if not ret:
        print("End of video reached. Breaking loop.")
        break # End of video

    # Save frame as image
    frame_filename = os.path.join(output_folder, f"frame_{frame_count:03d}.jpg")
    cv2.imwrite(frame_filename, frame)

    # This print statement is the confirmation output you were looking for:
    print(f"Saved {frame_filename}")

    frame_count += 1

# Release resources
cap.release()
# Note: cv2.destroyAllWindows() is often unnecessary in non-GUI scripts
# running in environments like Colab or standard terminals.

print(f"\nExtraction complete. Total frames saved: {frame_count}")
```

```python
import pandas as pd

import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression

#Step 1: Create dataset with Experience and Salary

data = {

'Experience': [1, 2, 3, 4, 5, 6, 8, 10],

'Salary': [25000, 30000, 35000, 40000,45000, 50000, 60000, 70000]

}
df = pd.DataFrame(data)

#Step 2: Define features (X) and target (y)



x = df[[ 'Experience']]

y = df['Salary']

#Step 3: Train Linear Regression model

model = LinearRegression()

model.fit(x, y)

# Step 4: Predict salary for 7 years of experience

experience_input = [[7]]

predicted_salary = model.predict(experience_input)

print (f"Predicted salary for 7 years experience: {predicted_salary[0]:.2f}")

#Step 5: Visualization

plt.scatter(x, y, color='blue', label='Actual Data') # Scatter plot of actual data

plt.plot(x, model.predict(x), color='red', label='Regression Line') # Regression line

plt.scatter(7, predicted_salary, color='green', marker='o', s=100, label='Predicted (7 yrs)') #a Predicted point

plt.title("Salary Prediction based on Experience")

plt.xlabel("Years of Experience")

plt.ylabel("Salary (t)")

plt.legend()

plt.grid(True)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Step 1: Create dataset with Experience and Salary
data = {
'Experience': [1, 2, 3, 4, 5, 6, 8, 10, 12],
'Salary': [25000, 30000, 35000, 40000,45000, 50000, 60000, 70000, 80000]
}
df = pd.DataFrame(data)

# Step 2: Define features (X) and target (y)
x = df[['Experience']]
y = df['Salary']

# Step 3: Train Linear Regression model
model = LinearRegression()
model.fit(x, y)

# Step 4: Predict salary for 11 years of experience (modified)
experience_input = [[11]]
predicted_salary = model.predict(experience_input)

print(f"Predicted salary for 11 years experience: {predicted_salary[0]:.2f}")

# Step 5: Visualization (modified to highlight the new prediction)
plt.scatter(x, y, color='blue', label='Actual Data') # Scatter plot of actual data
plt.plot(x, model.predict(x), color='red', label='Regression Line') # Regression line
plt.scatter(11, predicted_salary, color='green', marker='o', s=100, label='Predicted (11 yrs)') # The new predicted point

plt.title("Salary Prediction based on Experience")
plt.xlabel("Years of Experience")
plt.ylabel("Salary (t)")
plt.legend()
plt.grid(True)
plt.show()
```
