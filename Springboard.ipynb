```python
#Stemming
from nltk.stem import PorterStemmer
#Create a Porter Stemmer instance
porter_stemmer=PorterStemmer()
#Example words for stemming
words=["running","jumps","happily","running","happily"]
#Apply stemming to each word
stemmed_words=[porter_stemmer.stem(word) for word in words]
#Print the results
print("Original words:",words)
print("Stemmed words:",stemmed_words)
```

```python
#Stemming using text file
# Import the PorterStemmer
from nltk.stem import PorterStemmer

# Create a Porter Stemmer instance
porter_stemmer = PorterStemmer()

# Read data from a text file
with open("input.txt", "r") as file:
    text = file.read()

# Split text into words
words = text.split()

# Apply stemming
stemmed_words = [porter_stemmer.stem(word) for word in words]

# Print results
print("Original words:", words)
print("Stemmed words:", stemmed_words)

# Optionally save the stemmed output to another file
with open("stemmed_output.txt", "w") as file:
    file.write(" ".join(stemmed_words))
```

```python
#Anamoly detection
#Sample transactions: (amount, loctaion)
transactions= [
    (20,"local"), #grocery
    (50,"local"), #gas
    (1000,"local"), #online shopping
    (500,"foreign"), #jewelry purchase abroad
    (200,"local"), #electronics
]
#Define threshlods
MAX_NORMAL_AMOUNT=1000 #anything above this is suspicious
ALLOWED_LOCATIONS=["local"]

def detect_anomalies(transactions):
    anomalies = []
    for idx, (amount, location) in enumerate(transactions, start=1):

        # check for unusually high amount or unusual location
        if amount > MAX_NORMAL_AMOUNT or location not in ALLOWED_LOCATIONS:
            anomalies.append({
                "transaction_id": idx,
                "amount": amount,
                "location": location,
                "reason": "High amount" if amount > MAX_NORMAL_AMOUNT else "Unusual location"
            })

    return anomalies

# Run detection
anomalies = detect_anomalies(transactions)

# Print results
if anomalies:
    print("Suspicious transactions detected:")
    for anomaly in anomalies:
        print(f"ID: {anomaly['transaction_id']} : $ "
              f"Amount = {anomaly['amount']} | "
              f"Location = {anomaly['location']} | "
              f"{anomaly['reason']}")
else:
    print("No anomalies detected.")
```

```python
class ChatBot:
    def __init__(self):
        # Group questions that should have the same answer
        self.responses = {

            # 1. Name related questions
            "My name is ChatBot!": [
                "what is your name",
                "what's your name",
                "who are you",
                "tell me your name",
                "your name"
            ],

            # 2. Feeling / how are you questions
            "I'm doing great, thanks!": [
                "how are you",
                "how are you doing",
                "how do you feel",
                "are you ok",
                "how's it going"
            ],

            # 3. Purpose related questions
            "I'm a chatbot created to help you!": [
                "what are you",
                "what do you do",
                "what is your purpose",
                "why do you exist"
            ],

            # 4. Creator related questions
            "I was created by a Python developer!": [
                "who created you",
                "who made you",
                "who developed you"
            ],

            # 5. Weather questions
            "I don't have senses, but I hope the weather is nice where you are!": [
                "how is the weather",
                "what's the weather",
                "is it raining",
                "is it hot today"
            ],

            # 6. Joke questions
            "Here’s a joke: Why do programmers hate nature? Too many bugs!": [
                "tell me a joke",
                "say a joke",
                "make me laugh",
                "joke"
            ],

            # 7. Time questions
            "I don’t keep track of time, but you can check your clock!": [
                "what's the time",
                "tell me the time",
                "time please"
            ],

            # 8. About Python
            "Python is a powerful programming language used for AI, ML, and web development!": [
                "what is python",
                "tell me about python",
                "python language"
            ],

            # 9. Gratitude response
            "You're welcome! Happy to help 😊": [
                "thank you",
                "thanks",
                "thank u"
            ],
        }

    def get_response(self, user_input):
        user_input = user_input.lower().strip()

        # Search in grouped responses
        for response, questions in self.responses.items():
            if user_input in questions:
                return response

        return "Sorry, I don't understand."

# -------- RUN THE CHATBOT ---------

bot = ChatBot()
print("Chatbot: Hello! Type 'bye' to exit.")

while True:
    user = input("You: ").strip().lower()

    if user == "bye":
        print("Chatbot: Goodbye! Have a great day!")
        break

    print("Chatbot:", bot.get_response(user))
```

```python
# Install required packages
!apt-get -y install ffmpeg
!pip install openai-whisper langdetect pydub

import whisper
from langdetect import detect, LangDetectException
from pydub import AudioSegment

# -----------------------------------------------------
# STEP 1: Convert MP3 to WAV
# -----------------------------------------------------
audio_path = "/content/telugu_dialogue.mp3"   # <-- upload your MP3 to Colab first
wav_path = "/content/telugu_dialogue.wav"

# Convert MP3 → WAV
AudioSegment.from_file(audio_path, format="mp3").export(wav_path, format="wav")
print("MP3 converted to WAV successfully!")

# -----------------------------------------------------
# STEP 2: Load Whisper model
# -----------------------------------------------------
print("Loading Whisper model... (takes ~20 seconds)")
model = whisper.load_model("small")   # "base" or "tiny" works on CPU, "small" needs GPU

# -----------------------------------------------------
# STEP 3: Transcribe audio
# -----------------------------------------------------
print("Transcribing...")
result = model.transcribe(wav_path)
text = result["text"].strip()

print("\n---------------------------")
print("TRANSCRIBED TEXT:\n")
print(text)
print("\n---------------------------")

# -----------------------------------------------------
# STEP 4 (Optional): Detect Language
# -----------------------------------------------------
try:
    detected_lang = detect(text)
    print("Detected Language:", detected_lang)
except LangDetectException:
    print("Language detection failed (text too short).")
```
