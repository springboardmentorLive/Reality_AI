```python
class chatbot:
  def __init__(self):

    self.responses = {
        "My name is surya!":[
            "what is your name?",
            "what's your name?",
            "who are you?",
            "tell me your name?",
            "your name please?"

        ],
        "I am doing great thank you":
        [
            "how are you?",
            "how are you doing?",
            "how are you feeling?",
            "how are you today?",
            "how's it going?"
        ],
        "I'm a chatbot created to help you with your queries":
        [
            "what can you do?",
            "what can you help you with?",
            "what are your capabilities?",
            "why did you exist?"
        ],
        "Hi there!": [
            "hello",
            "hi",
            "hey"
        ],
        "I don't have an age.": [
            "how old are you?",
            "what is your age?"
        ],
        "I can provide information and chat with you.": [
            "what else can you do?",
            "what are your other functions?"
        ]
    }

  def get_response(self,user_input):
      user_input = user_input.lower().strip()

      for response,questions in self.responses.items():
        if any(question in user_input for question in questions):
          return response
      return"I'm sorry, I didn't understand that."
```

```python
bot = chatbot()
print(bot.get_response("what's your name?"))
print(bot.get_response("how are you?"))
print(bot.get_response("what can you do?"))
print(bot.get_response("hello"))
print(bot.get_response("how old are you?"))
print(bot.get_response("what else can you do?"))
print(bot.get_response("tell me about yourself"))
```

```python
print("chatbot : Hi! I am HealthCareBot. Type 'bye' to exit.")

while True:
    user = input("you: ").lower()

    # Basic greetings
    if user == "hi" or user == "hello":
        print("chatbot: Hello! How can I help you with your health today?")

    elif user == "how are you?":
        print("chatbot: I am functioning well! How are you feeling today?")

    # Common symptoms
    elif "fever" in user:
        print("chatbot: You may have a fever. Drink plenty of water and take rest.")
        print("chatbot: If your fever lasts more than 2 days, please visit a doctor.")

    elif "headache" in user:
        print("chatbot: Headaches can be caused by stress or dehydration.")
        print("chatbot: Try drinking water and relaxing for a while.")

    elif "cough" in user:
        print("chatbot: A cough may be due to cold or throat infection.")
        print("chatbot: If it lasts more than a week, consult a healthcare provider.")

    elif "cold" in user or "running nose" in user:
        print("chatbot: You might be experiencing a common cold.")
        print("chatbot: Warm water and steam inhalation can give relief.")

    elif "bp" in user or "blood pressure" in user:
        print("chatbot: Normal BP is around 120/80.")
        print("chatbot: If it's too high or low, please visit a doctor.")

    elif "sugar" in user or "diabetes" in user:
        print("chatbot: Keep your sugar levels in check with regular testing.")
        print("chatbot: Follow a balanced diet and consult your doctor regularly.")

    elif "medicine" in user or "tablet" in user:
        print("chatbot: I can give general advice only.")
        print("chatbot: For exact medicine or dosage, please consult a doctor.")

    # Exit
    elif user == "bye":
        print("chatbot: Goodbye! Take care of your health.")
        break

    # Default fallback
    else:
        print("chatbot: I'm not sure about that. Please ask health-related questions.")
```

```python
from groq import Groq

client = Groq(api_key = "")

def ask_bot(prompt):
	response = client.chat.completions.create(
		model = "llama-3.3-70b-Versatile",
		messages =[{"role":"user","content":prompt}])

	return response.choices[0].message.content # Corrected 'choice' to 'choices'
print("AI Chatbot Ready!type 'quit' to exit.\n")

while True:
	user = input("You:")
	if user.lower() =="quit":
		print("Bot: bye!")
		break
	print("Bot:", ask_bot(user))
```

```python
from groq import Groq

client = Groq(api_key="")

def ask_health_bot(prompt):
    health_prompt = f"""User Question: {prompt} """
    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[{"role": "user", "content": health_prompt}]
    )
    return response.choices[0].message.content


print(" HealthCare AI Chatbot Ready! Type 'quit' to exit.\n")

while True:
    user = input("You: ")
    if user.lower() == "quit":
        print("Bot: Take care! Goodbye.")
        break
    print("Bot:", ask_health_bot(user))
```
