```python
import pandas as pd

left = pd.DataFrame (

{

"key": ["K0", "K1", "K2", "K3"], "Α": ["Α0", "A1", "A2", "Аз"], "B": ["B0", "B1", "B2", "B3"],

}

)

right = pd.DataFrame(

{

"key": ["K0", "K1", "K2", "K3"], "C": ["C0", "C1", "C2", "C3"], "D": ["D0", "D1", "D2", "D3"],

}

)

result = pd.merge(left, right, on="key")

result
```

```python
import pandas as pd
left = pd.DataFrame(
    {
        "key1": ["K0","K1", "K2", "K3"],
        "key2": ["K0", "K1", "K2", "K3"],
        "A": ["A0", "A1", "A2", "A3"],
        "B": ["B0", "B1", "B2", "B3"],
    }
)

right = pd.DataFrame(
    {
        "key1": ["K0", "K1", "K2", "K3"],
        "key2": ["K0", "K1", "K2", "K3"],
        "C": ["C1", "C2", "C3", "C4"],
        "D": ["D1", "D2", "D3", "D4"],
    }
)

result = pd.merge(left, right, how="left", on=["key1", "key2"])
result
```

```python
result = pd.merge(left, right, how="right", on=["key1", "key2"])
result
```

```python
result = pd.merge(left, right, how="inner", on=["key1", "key2"])
result
```

```python
result = pd.merge(left, right, how="outer", on=["key1", "key2"])
result
```

```python
import pandas as pd
data = {
    'Item_Name': ['Laptop', 'Book', 'Headphones', 'Backpack'],
    'Price': [55000, 499, None, 1200],
    'Store_Location': ['Mumbai', None, 'Bangalore', 'Delhi'],
    'Category': ['Electronics', 'Stationery', 'Electronics', 'Accessories']
}
print(data)
```

```python
df = pd.DataFrame(data)
print(df)
```

Step-by-Step Data Cleaning Method
1. Check for Missing Data:
Shows which cells are missing(True)


```python
df.isna()  #it is used to check the null values in the dataframe
```

```python
df.isna().sum() #It is like an extension of isna() where we get the sum
```

2. Drop Missing Data

```python
df.dropna()  #it is used to drop the rows with null values and the output only shows the rows with zero null values
```

Drop Columns with Nan/None or Null values

```python
df.dropna(axis = 1)
```

Drop Rows with Null values (default)


```python
df.dropna(axis = 0)
```

3. Drop Rows with missing values

```python
df.dropna(subset=['Store_Location'])
```

```python
df.at[3,'Price']=2000 #at method is used to update a particuar cell
df                            #at[index_no., Col_name] = new value
```

4. Fill Missing values

```python
df.fillna('0')  #fillna is used to fill values wherever null values are present
```

Rename Columns

```python
df.rename(columns={'Price': 'Cost'}, inplace=True)
df
```

```python
df['Item_Name'] =df['Item_Name'].str.upper()
df
```

```python
df['Category'] = df['Category'].str.lower()
df
```

5. Remove Outliers

```python
df = df[df['Cost'] < 10000]
```
