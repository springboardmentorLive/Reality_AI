```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
data = {
    'Experience': [1, 2, 3, 4, 5, 6, 8, 10],
    'Salary': [25000, 30000, 35000, 40000, 45000, 50000, 60000, 70000]
}
df = pd.DataFrame(data)
```

```python
X = df[['Experience']]
y = df['Salary']

model = LinearRegression()
model.fit(X, y)
```

```python
experience_input = [[7]]
predicted_salary = model.predict(experience_input)
print("Predicted Salary for 7 years of experience:", predicted_salary[0])
```

```python
plt.scatter(X, y, color='blue', label='Actual Data')
plt.plot(X, model.predict(X), color='red', label='Regression Line')
plt.scatter(7, predicted_salary, color='green', marker='o', s=100,
            label='Predicted Salary (7 yrs)')

plt.title("Salary Prediction based on Experience")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.legend()
plt.grid(True)
plt.show()

```
