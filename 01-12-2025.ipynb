```python
import pandas as pd

data = {"Name": ["Riya", "Karan", None, "Meena", "Amit",None,"Sneha","Vikas", "Neha", "Tina"],
        "Age": [28, 30, 22, None, 26, 21, None, 24, 29, None],
        "Gender": ['F', 'M', 'M', 'F', 'M', 'M', 'F', 'M', 'F', 'F'],
        "Score":[91, None, 76, 84, 89, 95, None, 88, 93, 90]}
df = pd.DataFrame(data)
df.to_csv("raw_data.csv", index=False)
```

```python
df_raw = pd.read_csv("raw_data.csv", nrows=10)
print("Original DataFrame (first 10 rows):\n")
print(df_raw)

df_cleaned = df_raw.dropna()
print("\nDataFrame after dropping rows with missing values:\n")
print(df_cleaned)
```

```python
df_full = pd.read_csv("raw_data.csv")
print("Original DataFrame (all rows):")
print(df_full)

df_cleaned_full = df_full.dropna()
print("\nDataFrame after dropping rows with missing values (all rows):")
print(df_cleaned_full)
```

```python
missing_values_count = df_full.isnull().sum()
print("Missing values count per column:")
print(missing_values_count)
```
