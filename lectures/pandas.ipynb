### Introduction to Pandas Library

Creating Pandas Dataframes

```python
import pandas as pd
left = pd.DataFrame(
    {
        "key":["K0", "K1", "K2", "AЗ"],
        "Α": ["Α0", "Α1", "A2", "A3"],
        "B": ["B0", "B1", "B2", "B3"],
    }
)

right = pd.DataFrame(
    {
       "key": ["K0", "К1", "K2", "К3"],
       "C": ["C0", "C1", "C2", "C3"],
       "D": ["D0", "D1", "D2", "D3"],
    }
)

result = pd.merge(left, right, on="key")

result
```

### Merging two DataFrames on a single key

```python
left=pd.DataFrame(
{
"key": ["K0", "K1", "K2", "A3"],
"A": ["A0", "A1", "A2", "A3"],
"B":["B0", "B1", "B2", "B3"],
}
)

right=pd.DataFrame(
{
"key": ["K0", "K1", "K2", "K3"],
"C": ["C0", "C1", "C2", "C3"],
"D": ["D0", "D1", "D2", "D3"],
}
)

result = pd.merge(left, right, on="key")

result
```

```python
left=pd.DataFrame(
{
"key1": ["K0", "K0", "K1", "K2"],
"key2": ["K0", "K1", "K0", "К1"],
"A": ["A0", "A1", "A2", "A3"],
"B":["B0", "B1", "B2", "B3"],
}
)

right=pd.DataFrame(
{
"key1": ["K0", "K1", "K1", "K2"],
"key2": ["K0", "K0", "K0", "K0"],
"C": ["C0", "C1", "C2", "C3"],
"D": ["D0", "D1", "D2", "D3"],
}
)

result = pd.merge(left, right, how="left", on=["key1", "key2"])

result
```

```python
result = pd.merge(left, right, how="right", on=["key1", "key2"])
result
```

```python
result = pd.merge(left, right, how="outer", on=["key1", "key2"])
result
```

```python
result = pd.merge(left, right, how="inner", on=["key1", "key2"])
result
```

```python
data = {'Name': ['Amal', 'John', 'Adinath', 'Megha'],
        'Age': [25, None, 30, None],
        'Gender': ['M', 'M', None, 'F']}


df = pd.DataFrame(data)
print(df)
```

### Detecting Null values
The pandas.isna() function is used to detect missing or null values (represented as NaN, None, or NaT)

```python
df.isna()
```

### Drop Rows with Any NaN

```python
df.dropna()
```

### Drop Rows with Missing in Specific Column

```python
 df.dropna(subset=['Age'])
```

### Filling the missing values with mean

```python
# Fill missing Age values with the mean
df["Age"].fillna(int(df["Age"].mean()), inplace=True)

# Fill missing Gender values with the mode
df["Gender"].fillna(df["Gender"].mode()[0], inplace=True)
```

```python
df
```

#### Detect Duplicates

```python
df.duplicated()
```

### Rename Columns

```python
df.rename(columns={'Name': 'Full Name'}, inplace=True)
df
```

### Standardize Text

```python
df['Full Name'] = df['Full Name'].str.upper()
df
```

### Filtering texts in pandas

```python
df2 = df[df["Gender"] == "M"]
df2
```
