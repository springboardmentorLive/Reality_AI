### Natural Language Processing using NLTK


```python
from nltk.stem import WordNetLemmatizer

import nltk

nltk.download('wordnet')

lemmatizer = WordNetLemmatizer()

print("rocks:", lemmatizer.lemmatize("rocks"))

print("corpora :", lemmatizer.lemmatize("corpora"))

# a denotes adjective in "pos"

print("better:", lemmatizer.lemmatize("better", pos="a"))
```

```python
from nltk.stem import PorterStemmer

# Create a Porter Stemmer instance

porter_stemmer = PorterStemmer()

# Example words for stemming

words = ["running", "jumps", "happily", "running", "happily"]

# Apply stemming to each word

stemmed_words = [porter_stemmer.stem(word) for word in words]

# Print the results

print("Original words:", words)

print("Stemmed words:", stemmed_words)
```

```python
from nltk.stem import PorterStemmer

#Create a PorterStemmer instance

porter_stemmer = PorterStemmer()

# Read words from a text file

with open("word.txt", "r") as file:
  text = file.read()

# Split words by space or newline

words = text.split()

# Apply stemming

stemmed_words = [porter_stemmer.stem(word) for word in words]

#Print results

print("Original words:", words)

print("Stemmed words:", stemmed_words)
```

```python
# Simple anomaly detection for credit card transactions

#Sample transactions: (amount, location)

transactions = [

(20, "local"),# grocery
(50, "local"),#gas
(1000, "local"),#online shopping
(508, "foreign"),# jewelry purchase abroad
(200, "local"),#electronics
]

#Define thresholds

MAX_NORMAL_AMOUNT = 1000 #anything above this is suspicious

ALLOWED_LOCATIONS = ["local"]

def detect_anomalies (transactions):
  anomalies = []
  for idx, (amount, location) in enumerate (transactions, start=1):
    if amount > MAX_NORMAL_AMOUNT or location not in ALLOWED_LOCATIONS:
      anomalies.append({
          "transaction_id":idx,
          "amount": amount,
          "location": location,
          "reason": "High amount" if amount> MAX_NORMAL_AMOUNT else "Unusual location"
      })
  return anomalies

#Run detection
anomalies = detect_anomalies(transactions)
#Print results
if anomalies:
  print("Suspicious transactions detected:")
  for anomaly in anomalies:
    print(f"ID {anomaly['transaction_id']}: ${anomaly['amount']} at {anomaly['location']} {anomaly['reason']}")
else:
  print("No anomalies detected.")
```

### Simple static chatbot

```python
print("Chatbot: Hi! I am your chatbot. Type 'bye' to exit.")

while True:
    user = input("You: ")

    if user.lower() == "hi":
        print("Chatbot: Hello! How can I help you?")

    elif user.lower() == "how are you":
        print("Chatbot: I am fine! Thank you ")

    elif user.lower() == "bye":
        print("Chatbot: Goodbye! Have a nice day!")
        break

    else:
        print("Chatbot: Sorry, I don't understand.")
```

## Advanced Hardcoded chatbot

```python
import re

class ChatBot:
    def __init__(self):
        # Intent-based responses
        self.intents = {
            "name": {
                "response": "My name is ChatBot.",
                "patterns": [
                    "what is your name",
                    "who are you",
                    "tell me your name",
                    "your name",
                    "identify yourself"
                ]
            },
            "status": {
                "response": "I'm doing great, thank you for asking.",
                "patterns": [
                    "how are you",
                    "how are you doing",
                    "how do you feel",
                    "are you okay",
                    "how is it going"
                ]
            },
            "purpose": {
                "response": "I am a chatbot designed to assist and answer your questions.",
                "patterns": [
                    "what are you",
                    "what do you do",
                    "what is your purpose",
                    "why do you exist"
                ]
            }
        }

        self.threshold = 0.3  # similarity threshold

    def normalize(self, text):
        text = text.lower()
        text = re.sub(r"[^a-z\s]", "", text)
        return set(text.split())

    def similarity(self, set1, set2):
        # Jaccard similarity
        return len(set1 & set2) / len(set1 | set2)

    def get_response(self, user_input):
        user_tokens = self.normalize(user_input)

        best_score = 0
        best_response = None

        for intent in self.intents.values():
            for pattern in intent["patterns"]:
                pattern_tokens = self.normalize(pattern)
                score = self.similarity(user_tokens, pattern_tokens)

                if score > best_score:
                    best_score = score
                    best_response = intent["response"]

        if best_score >= self.threshold:
            return best_response

        return "I'm not confident enough to answer that. Could you rephrase?"

# ------------------
# Usage
# ------------------
bot = ChatBot()

print(bot.get_response("What's your name?"))
print(bot.get_response("who exactly are you"))
print(bot.get_response("How are you"))
print(bot.get_response("how's life going"))
```
