```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import style

time_data = {
    'facebook': 120,
    'youtube': 90,
    'instagram': 40,
    'snapchat': 28
}

social_media = list(time_data.keys())
time_speed = list(time_data.values())

plt.bar(social_media, time_speed)
plt.xlabel("Social Media Platform")
plt.ylabel("Time Spent (minutes)")
plt.title("Time Spent on Different Social Media Apps")
plt.show()
```

```python
style.use('ggplot')

social_media = ['facebook', 'youtube', 'instagram', 'snapchat']
time_spent = [120, 90, 40, 28]

colors = ['red', 'yellow', 'blue', 'pink']

plt.bar(social_media, time_spent, color=colors)
plt.xlabel('social media')
plt.ylabel('time')
plt.title('average time speed on social media')
plt.show()
```

```python
plt.figure(figsize=(8, 8))
plt.pie(time_spent, labels=social_media, autopct='%1.1f%%', startangle=90, colors=['red', 'yellow', 'blue', 'pink'])
plt.title('Time Spent on Social Media Apps')
plt.axis('equal')
plt.show()
```

```python
import csv

social_media = []
time_spent = []

csv_filename = 'social_media_time_spent.csv'

with open(csv_filename, mode='r') as file:
    reader = csv.reader(file)
    header = next(reader)
    for row in reader:
        social_media.append(row[0])
        time_spent.append(int(row[1]))

plt.figure(figsize=(8, 8))
plt.pie(time_spent, labels=social_media, autopct='%1.1f%%', startangle=90, colors=['red', 'yellow', 'blue', 'pink'])
plt.title('Time Spent on Social Media Apps')
plt.axis('equal')
plt.show()
```

```python
import pandas as pd

data = {
    "Name": ['Riya', 'Karan', None, 'Meena', 'Amit', None, 'Sneha', 'Vikas', 'Neha', 'Tina'],
    "Age":  [28, 30, 22, None, 26, 21, None, 24, 29, None],
    "Gender": ['F', 'M', 'M', 'F', 'M', 'M', 'F', 'M', 'F', 'F'],
    'Score': [91, None, 76, 84, 89, 95, None, 88, 93, 90]
}

df_original = pd.DataFrame(data)

df_original.to_csv('raw_data.csv')

print("DataFrame created and saved to 'raw_data.csv' successfully.")
```

```python
import cv2
import os
from PIL import Image
from IPython.display import display

video_path = "/content/AIvideo.mp4"
output_folder = "/content/frames"
os.makedirs(output_folder, exist_ok=True)

cap = cv2.VideoCapture(video_path)
if not cap.isOpened():
    raise IOError("Error: Could not open video.")

frame_count = 0
frames_to_display = 5

while True:
    ret, frame = cap.read()
    if not ret or frame_count >= frames_to_display:
        break

    frame_filename = os.path.join(output_folder, f"frame_{frame_count:03d}.jpg")
    cv2.imwrite(frame_filename, frame)
    print(f"Saved: {frame_filename}")

    # Open with PIL and explicitly display
    pil_img = Image.open(frame_filename)
    display(pil_img)

    frame_count += 1

cap.release()
cv2.destroyAllWindows()
print(f"\nTotal frames extracted and displayed: {frame_count}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Your data
data = {
    'Experience': [1, 2, 3, 4, 5, 6, 8, 10],
    'Salary':     [25000, 30000, 35000, 40000, 45000, 50000, 60000, 70000]
}

# Convert lists to numpy arrays
X = np.array(data['Experience']).reshape(-1, 1)
y = np.array(data['Salary'])

# Create and train linear regression model
model = LinearRegression()
model.fit(X, y)

# Predict salaries for regression line
y_pred = model.predict(X)

# Plotting
plt.figure(figsize=(8,5))
plt.scatter(X, y, color='blue', label='Actual Data')
plt.plot(X, y_pred, color='red', linewidth=2, label='Regression Line')

plt.xlabel("Experience (years)")
plt.ylabel("Salary")
plt.title("Experience vs Salary - Linear Regression")
plt.legend()
plt.grid(True)
plt.show()
```

```python
exp = np.array([[11]])
pred_salary = model.predict(exp)

print("Predicted Salary for 7 years experience:", pred_salary[0])
```
