```python
from nltk.stem import WordNetLemmatizer
import nltk
nltk.download('wordnet')
lemmatizer = WordNetLemmatizer()

print("rocks :", lemmatizer.lemmatize("rocks"))
print("corpora :", lemmatizer.lemmatize("corpora"))


print("better :", lemmatizer.lemmatize("better", pos ="a"))

```

```python
from nltk.stem import PorterStemmer

porter_stemmer = PorterStemmer()

words = ["running" , "jumps","happily","running","happyliy"]

stemmed_word = [porter_stemmer.stem(word) for word in words]

print("orignal words:",words)
print("stemmed words:",stemmed_word)
```

```python
from zipfile import MAX_EXTRACT_VERSION
transactions = [
    (50, "local"),
    (100, "international"),
    (75, "local"),
    (500, "international"),
    (25, "local"),
    (1000, "international"),
    (150, "local"),
    (300, "international"),
    (80, "local"),
    (5000, "international")
]

MAX_EXTRACT_AMOUNT = 500
ALLOWED_LoCATION = ["international"]
def detect_anomalies(transactions):
  anomalies = []
  for idx , (amount,location) in enumerate(transactions):
    if amount > MAX_EXTRACT_AMOUNT or location not in ALLOWED_LoCATION:
      anomalies.append({
          "transaction_id":idx,
          "amount":amount,
          "location":location,
          "reason":"hight amount" if amount > MAX_EXTRACT_AMOUNT else "invalid location"
      })
  return anomalies

anomalies = detect_anomalies(transactions)

if anomalies:
  print("anomalies detected:")
  for anomaly in anomalies:
    print(f"transaction ID:{anomaly['transaction_id']}:${(anomaly['amount'])} at {anomaly['location']} -> {anomaly['reason']}")
else:
  print("no anomalies detected")
```
