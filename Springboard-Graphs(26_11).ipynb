```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import style
```

```python
time_data = {'cricket': 150, 'football': 95, 'badminton': 105, 'swimming': 70}

sports = list(time_data.keys())
time_spent = list(time_data.values())

plt.bar(sports, time_spent, color=['red', 'blue', 'green', 'orange'], width=0.5)
plt.xlabel('Sports')
plt.ylabel('Time Spent')
plt.title('Sports vs Time Spent')
style.use('ggplot')
plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib import style

time_data = {'cricket': 150, 'football': 95, 'badminton': 105, 'swimming': 70}

sports = list(time_data.keys())
time_spent = list(time_data.values())
colors = ['red', 'blue', 'green', 'orange']

bars = plt.bar(sports, time_spent, color=colors, width=0.5)

plt.legend(bars, sports, loc="upper right")

plt.xlabel('Sports')
plt.ylabel('Time Spent')
plt.title('Sports vs Time Spent')
style.use('ggplot')
plt.show()
```

```python
import matplotlib.pyplot as plt

time_data = {'cricket': 150, 'football': 95, 'badminton': 105, 'swimming': 70}
colors = ['red', 'blue', 'green', 'orange']
sports = list(time_data.keys())
time_spent = list(time_data.values())

plt.pie(time_spent, labels=sports,colors = colors, autopct='%1.1f%%')
plt.title("Time Spent on Sports")
plt.show()
```

```python
import matplotlib.pyplot as plt
import csv

Authors = []
Counts = {}

with open('books.csv', 'r', encoding='latin1') as file:
    data = csv.reader(file, delimiter=';')
    next(data)

    for row in data:
        author = row[2]
        Counts[author] = Counts.get(author, 0) + 1

top5 = sorted(Counts.items(), key=lambda x: x[1], reverse=True)[:5]

Authors = [a for a, c in top5]
Books = [c for a, c in top5]
colors = ['purple','green','orange','blue','yellow']

plt.pie(Books, labels=Authors,colors = colors, autopct='%.2f%%')
plt.title("Top 5 Authors")
plt.show()
```
