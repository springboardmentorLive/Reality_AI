```python
import pandas as pd
data = {'Name': ['Geethika', 'Priya', 'Lakshmi', 'Sarath'],
        'Age': [25, None, 30, None],
        'Gender': ['F', 'F', None, 'M']}
print(data)
```

```python
df = pd.DataFrame(data)
print(df)
```

Step-by-Step Data Cleaning Methods

Check for Missing Data

Shows which cells are missing (True)
Summarizes how many missing values per column -- isna()

```python
df.isna()
```

```python
df.isna()
```

2. Drop Missing Data 🔹 Drop Rows with Any NaN

```python
df.dropna()
```

 Drop Columns with Any NaN

```python
df.dropna(axis=1)
```

 Drop Rows with Missing in Specific Column

```python
 df.dropna(subset=['Age'])
```

```python
df.at[1,'Gender']='F'
df
```

Fill Missing Data (Imputation) 🔹 Fill with a Constan

```python
df.fillna(0)
```

Mean/Median/Mode

```python
df['Age'].fillna(df['Age'].mean(), inplace=True)
df['Gender'].fillna(df['Gender'].mode()[0], inplace=True)
```

Detect Duplicates

```python
df.duplicated()
#df.drop_duplicates()
```

```python
df.drop_duplicates()
```

Convert Data Types

```python
df['Age'] = df['Age'].astype(int)
```

6. Rename Columns

```python
df.rename(columns={'Name': 'FullName'}, inplace=True)
df
```

7. Standardize Text


```python
df['FullName'] = df['FullName'].str.upper()
df
```

8.Remove Outliers (Optional)

```python
df = df[df['Age'] < 100]  # Example filter
```
