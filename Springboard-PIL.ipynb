Adding Text on the Image

```python
# Importing the PIL(Python Imaging Library) Library
from PIL import Image
from PIL import ImageDraw

# Open an Image
img = Image.open('/content/bulbon.jpg')

# Call draw Method to add 2d graphics in an image
I1 = ImageDraw.Draw(img)

# Add text to an image
I1.text((10,16), "Bulb is glowing", fill=(105, 100, 0))

# Display edited image
img.show()

# Save the edited image
img.save("bulb.png")
```

```python
from PIL import Image

# Open an image
image = Image.open('/content/bulbon.jpg')

# Display the image
image.show()

# Resize the image
resized_image = image.resize((200, 200))

# Save the image
resized_image.save("resized_eg.jpg")
```
