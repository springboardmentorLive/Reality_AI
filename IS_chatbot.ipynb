```python
class ChatBot:

    def __init__(self):
        # Group new questions within the existing lists to receive the same output
        self.responses = {
            # Answer as key, list of questions as value
            "My name is ChatBot.": [
                "what is your name",
                "what's your name",
                "who are you",
                "tell me your name",
                "your name",
                "what are you called",
                "do you have a name",
                "identify yourself",
                "can you tell me who you are",
                "what should I call you"
            ],

            # Answer as key, list of questions as value
            "I am doing great, thanks for asking!": [
                "how are you",
                "how are you doing",
                "how do you feel",
                "are you ok",
                "how's it going",
                "are you feeling well",
                "how are things going",
                "is everything alright with you",
                "hope you are doing well"
            ],

            # Answer as key, list of questions as value
            "I am a chatbot created to help answer your questions.": [
                "what are you",
                "what do you do",
                "what is your purpose",
                "why do you exist",
                "what can you do for me",
                "how do you work",
                "what is your function",
                "why were you made",
                "explain your purpose"
            ]
        }

    def get_response(self, user_input):
        user_input = user_input.lower().strip()

        # Iterate through the dictionary where keys are answers and values are lists of questions
        for answer_text, questions_phrases in self.responses.items():
            for question_phrase in questions_phrases:
                # Check if the predefined phrase exists within the user's input
                if question_phrase in user_input:
                    return answer_text

        return "I'm not sure how to answer that."

# Usage Example (testing the new questions):
bot = ChatBot()
print(f"'What should I call you?' -> {bot.get_response('What should I call you?')}")
print(f"'do you have a name?' -> {bot.get_response('do you have a name?')}")
print(f"'How are things going today?' -> {bot.get_response('How are things going today?')}")
print(f"'What can you do for me?' -> {bot.get_response('What can you do for me?')}")
```
