```python
df = df.drop_duplicates()
df.isnull().sum()

```

```python
df = df.fillna(0)
df = df.dropna()
```

```python
len(df)
```

```python
df.groupby("Party")["Total Votes"].sum()
```

```python
df["Total Votes"] = pd.to_numeric(df["Total Votes"], errors="coerce")
```

```python
winners = df.loc[df.groupby("Party")["Total Votes"].idxmax()]
winners.head()
```

```python
winners["Party"].value_counts()
```

```python
import matplotlib.pyplot as plt

df.groupby("Party")["Total Votes"].sum().plot(kind="bar")
plt.title("Party-wise Total Votes")
plt.ylabel("Total Votes")
plt.show()
```

```python
winners["Party"].value_counts().plot(kind="bar")
plt.title("Seats Won by Each Party")
plt.ylabel("Number of Seats")
plt.show()
```
