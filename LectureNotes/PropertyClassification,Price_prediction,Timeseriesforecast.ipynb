```python
# Your code achieves a very basic property (house) condition classification using Machine Learning.
# In simple words:
# It takes a house image, measures how bright it is, and predicts whether the property is Old, Average, or New.
# That’s it.
# Meaning:
# You give the model examples:
# Brightness ≈ 100 → Old
# Brightness ≈ 150 → Average
# Brightness ≈ 200 → New
# The SVM learns this relationship.
# ⚠️ This is synthetic (dummy) training data, not real data.
# Reads the image in grayscale
# Each pixel has a value between 0–255
# Computes average pixel intensity
# Converts the entire image into one number
# This number becomes the input feature for ML.
```

```python

#PROPERTY CLASSIFICATION
import cv2
import numpy as np
from sklearn.svm import SVC

# Training data (features must be in a 2D array)
X_train = [[100], [150], [200]]
y_train = ["Old", "Average", "New"]  # Target labels

# Initialize and train the model
model = SVC()
model.fit(X_train, y_train)

# Test image: '8' loads as reduced grayscale
img = cv2.imread("/content/house.jpg", 8)

# Check if image was loaded correctly
if img is not None:
    # Calculate average brightness
    brightness = int(np.mean(img))

    # Predict: Input must be a 2D array [[value]]
    prediction = model.predict([[brightness]])

    print("Property Condition:", prediction[0])
else:
    print("Error: Could not find 'images/house.jpg'. Check the file path.")
```

```python
#PRICE PREDICTION
import pandas as pd
from sklearn.linear_model import LinearRegression
import os # Import os module to create directories and files

# Create a dummy data directory and file if they don't exist
if not os.path.exists('data'):
    os.makedirs('data')

csv_path = 'data/real_estate_prices.csv'
if not os.path.exists(csv_path):
    # Create a dummy CSV file with relevant columns
    dummy_data = {
        'year': [2018, 2019, 2020, 2021, 2022],
        'area': [1200, 1300, 1400, 1500, 1600],
        'rooms': [3, 3, 4, 4, 4],
        'price': [250000, 270000, 300000, 330000, 360000]
    }
    dummy_df = pd.DataFrame(dummy_data)
    dummy_df.to_csv(csv_path, index=False)
    print(f"Created dummy data file at {csv_path}")

data = pd.read_csv("data/real_estate_prices.csv")

X = data[['area', 'rooms']]
y = data['price']

model = LinearRegression()

model.fit(X, y)

# Fix: Convert new_house data to a DataFrame with column names matching X
new_house_data = pd.DataFrame([[1600, 4]], columns=['area', 'rooms'])

predicted_price = model.predict(new_house_data)

print("Predicted Property Price: ", int(predicted_price[0]))
```

```python
                                   #or
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression
import pandas as pd

# Generate synthetic regression dataset
X, y = make_regression(
    n_samples=100,
    n_features=2,
    noise=10,
    random_state=42
)

# Convert to DataFrame
data = pd.DataFrame(X, columns=["area", "rooms"])
data["price"] = y

# Train model
model = LinearRegression()
model.fit(data[["area", "rooms"]], data["price"])

# Predict
new_house = [[0.5, -1.2]]  # scaled values
predicted_price = model.predict(new_house)

print("Predicted Property Price:", predicted_price[0])
```
