```python
import pandas as pd
left=pd.DataFrame(
    {"key1":['k0','k1','k2','k3'],
    "key2":['k0','k1','k0','k1'],
     "A":['A0','A1','A2','A3'],
     "B":['B0','B1','B2','B3']
     }
)
right=pd.DataFrame(
    {'key1':['k0','k1','k2','k3'],
      'key2':['k0','k1','k0','k0'],
     'C':['C0','C1','C2','C3'],
     "D":['D0','D1','D2','D3']
     }
)
result=pd.merge(left,right,on=['key1','key2'])
result
```

```python
import pandas as pd
left=pd.DataFrame(
    {"key1":['k0','k1','k2','k3'],
    "key2":['k0','k1','k0','k1'],
     "A":['A0','A1','A2','A3'],
     "B":['B0','B1','B2','B3']
     }
)
right=pd.DataFrame(
    {'key1':['k0','k1','k2','k3'],
      'key2':['k0','k1','k0','k0'],
     'C':['C0','C1','C2','C3'],
     "D":['D0','D1','D2','D3']
     }
)
result1=pd.merge(left,right,how='left',on=['key1','key2'])
result1
```

```python

result2=pd.merge(left,right,how='right',on=['key1','key2'])
result2
```

```python

result3=pd.merge(left,right,how='outer',on=['key1','key2'])
result3
```

```python

result4=pd.merge(left,right,how='inner',on=['key1','key2'])
result4
```

1.Handling Missing Value**

```python
data={'name':['Abhiram','Arjun','Akshaya','sanju'],
      "Age":[25,None,30,None],
      "Gender":['M','M',None,'F']
}
print(data)
df=pd.DataFrame(data)
df
```

**finding missing values**

```python
print(df.isna())
print(df.isna().sum()) #in each column wise count
```

**Deleting missing values**

```python
print(df.dropna())
print(df.dropna(axis=1)) #columns
print(df.dropna(axis=0)) #rows
```

```python
df.dropna(subset=['Gender'],inplace=False) #by specific column
```

```python
df.at[1,'Age']=32.0       #updating elements
df
```

```python
df = df.fillna(0)   # filling missing values with zero and assigning back to df
df
```

```python
df.duplicated()
df
#df.drop_duplicated()
```

```python
df.rename(columns={'name':'Fullname'},inplace=True)   #Renaming the columns
df
```

```python
#using upper and lower methods
df['Fullname'] = df['Fullname'].str.upper()
print(df)
df['Gender']=df['Gender'].str.lower()
print(df)
```
