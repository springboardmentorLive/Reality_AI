```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import style
```

```python
time_data={'Facebook':50,'Instagram':90,'Whatsapp':64,'Twitter':54,'Snapchat':75,'Youtube':86}
social_media=list(time_data.keys())
time_speed=list(time_data.values())
plt.bar(social_media,time_speed,color='green',width=0.5)
plt.xlabel('social_media')
plt.ylabel('usage_time')
plt.title('Average time spending on SocialMedia')
```

```python
style.use('ggplot')
plt.figure(figsize=(6,5))
plt.bar(social_media,time_speed,color='orange',width=0.5)
plt.xlabel('social_media')
plt.ylabel('usage_time')
plt.title('Average time spending on SocialMedia')
```

```python
colors=['red','yellow','blue','purple','orange','skyblue']
style.use('ggplot')
plt.figure(figsize=(7,5))
bars=plt.bar(social_media,time_speed,color=colors,width=0.7)

# Correctly add a legend by passing both handles (bars) and labels (social_media)
plt.legend(bars, social_media, title='Social Media',loc='upper right')

plt.xlabel('social_media')
plt.ylabel('usage_time')
plt.title('Average time spending on SocialMedia')
plt.show()
```

```python
data_pie={'CSE':80,'ECE':60,'EEE':25,'CHEM':35,'MECH':40}
labels = list(data_pie.keys())
sizes = list(data_pie.values())
colors=['indigo','red','blue','orange','green']
plt.figure(figsize=(4,4))
plt.pie(sizes, labels=labels, autopct='%1.1f%%',colors=colors)
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.title('Department Distribution')
plt.legend(labels, title='Departments', loc='upper left', bbox_to_anchor=(1, 1)) # Add a legend
plt.show()
```

```python
import matplotlib.pyplot as plt
import csv
scores=[]
subjects=[]
with open('/content/book.csv','r') as csvfile:
  lines=csv.reader(csvfile,delimiter=',')
  next(lines) # Skip the header row
  for row in lines:
    subjects.append(row[0])
    scores.append(int(row[1]))
plt.pie(scores,labels=subjects,autopct='%.2f%%') # Corrected autopct format
plt.title('Marks of students')
plt.show()
```

**week4 -question on linear regression and visualisation**

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
data2={
    'Experience':[1,2,3,4,5,6,8,9,10],
    'Salary':[25000,30000,35000,40000,45000,50000,55000,60000,70000]
}
df2=pd.DataFrame(data2)
x=df2[['Experience']]
y=df2[['Salary']]
lr=LinearRegression()
lr.fit(x,y)
experience_input=[[7]]
# Convert experience_input to a DataFrame to match the format used during fitting
experience_input_df = pd.DataFrame(experience_input, columns=['Experience'])
predict_salary=lr.predict(experience_input_df)
print(f"Predicted salary for {experience_input[0][0]} years is ${predict_salary[0][0]:.2f}")

plt.scatter(x,y,color='blue',label='Actual Data')
plt.plot(x,lr.predict(x),color='red',label='Regression Line')
plt.scatter(7,predict_salary,color='green',marker='o',s=100,label='Predict Data')
plt.title('Salary Prediction based on Experience')
plt.xlabel('Experience')
plt.ylabel('Salary')
plt.grid(True)
plt.legend()
plt.show()
```
