```python
!pip install groq
from groq import Groq

client = Groq(api_key="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")

# Mode-based system instructions
MODES = {
    "normal": "You are a friendly general-purpose assistant.",
    "content": "You are a professional content creator. You write blogs, reels scripts, posts, captions, tweets, and creative content.",
    "story": "You are a storyteller. You write short stories, kids stories, emotional tales, and creative fiction.",
    "code": "You are a coding assistant. You explain code, write scripts, debug errors, and give optimized solutions.",
    "interview": "You are an interview trainer. You ask interview questions and evaluate answers strictly.",
    "jokes": "You are a funny comedy chatbot. You tell clean jokes and witty responses.",
}

current_mode = "normal"

def ask_bot(prompt, mode):
    system_prompt = MODES.get(mode, MODES["normal"])

    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": prompt}
        ]
    )

    return response.choices[0].message.content


print("🚀 Multi-Mode AI Chatbot Ready!")
print("Type '/help' to see available modes.\n")

while True:
    user = input("You: ")

    # Exit command
    if user.lower() == "quit":
        print("Bot: Bye! Have a great day!")
        break

    # Show commands
    if user.lower() == "/help":
        print("\nAvailable Modes:")
        for m in MODES.keys():
            print(f" - /{m}")
        print("Type '/normal' to return to default mode.\n")
        continue

    # Changing mode
    if user.startswith("/"):
        mode_name = user[1:].strip()
        if mode_name in MODES:
            current_mode = mode_name
            print(f"🔄 Mode switched to: {current_mode}\n")
        else:
            print("❌ Unknown mode. Type /help to see all modes.")
        continue
    # Process chat request
    bot_reply = ask_bot(user, current_mode)
    print("Bot:", bot_reply, "\n")
```
