```python
import cv2
import os
from IPython.display import Image, display

#  Path to your video file
video_path = "/content/Best nature whatsapp status__village nature__ video shorts nature viral trending.mp4"

#  Folder to save extracted frames
output_folder = "/content/frames"
os.makedirs(output_folder, exist_ok=True)

#  Load the video
cap = cv2.VideoCapture(video_path)

#  Check if video opened successfully
if not cap.isOpened():
    print("Error: Could not open video.")
    exit()

frame_count = 0
max_frames = 12  # Only extract 12 images

#  Loop through video and save first 12frames
while frame_count < max_frames:
    ret, frame = cap.read()
    if not ret:
        break  # End of video

    # Save frame as image
    frame_filename = os.path.join(output_folder, f"frame_{frame_count:03d}.jpg")
    cv2.imwrite(frame_filename, frame)
    print(f"Saved {frame_filename}")

    frame_count += 1

#  Release resources
cap.release()
cv2.destroyAllWindows()
print(" Extracted 10 frames successfully.")
# List saved images in the frames folder
print("\nSaved images:")
print(os.listdir(output_folder))

# Display all extracted frames
for i in range(12):
    frame_path = os.path.join(output_folder, f"frame_{i:03d}.jpg")
    if os.path.exists(frame_path):
        display(Image(filename=frame_path))
```

```python
#Drawing on image
from PIL import Image
from PIL import ImageDraw
import matplotlib.pyplot as plt

img2path="/content/download.jfif"
img2=Image.open(img2path)
plt.figure(figsize=(6,6))
plt.imshow(img2)
plt.title("Original Image")
plt.axis('off')
plt.show()

I1=ImageDraw.Draw(img2)                                      #Draw() method to add 2d graphics to image
I1.text((10,16)," WORK,That's how u get it",fill=(255,255,255))    #text() method a graphic ,fill is color of text,(10,16) is size of text

# Display the image AFTER drawing, not the ImageDraw object itself
plt.figure(figsize=(6,6))
plt.imshow(img2)
plt.title("Image with Text")
plt.axis('off')
plt.show()
```
