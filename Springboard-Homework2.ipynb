```python
# Import Libraries
import pandas as pd
import numpy as np
```

```python
# Create the dataset
data = {
    'Name': ['Riya', 'Karan', None, 'Meena', 'Amit', None, 'Sneha', 'Vikas', 'Neha', 'Tina'],
    'Age': [28, 30, 22, None, 26, 21, None, 24, 29, None],
    'Gender': ['F', 'M', 'M', 'F', 'M', 'M', 'F', 'M', 'F', 'F'],
    'Score': [91, None, 76, 84, 89, 95, None, 88, 93, 90]
}
df = pd.DataFrame(data)
```

```python
# Save it to csv
df.to_csv('raw_data.csv', index=False)
```

```python
# Read the first 5 rows of csv file
df_read = pd.read_csv('raw_data.csv')
print("First 5 rows of raw_data.csv:")
print(df_read.head(5))
```

```python
# Count the number of missing values in each column
print("Missing values in each column:")
print(df_read.isnull().sum())
```

```python
# drop rows with missing or null values
df_dropped = df_read.dropna()
print(df_dropped)
```

```python
print(df_dropped.isnull().sum())
```

```python
# create a copy of csv file to fill missing values
df_cleaned = df_read.copy()
```

```python
# Fill the missing names in 'Name' column with the help of name_fill_list
name_fill_list = ['Mowleen', 'Hari']
missing_name_indices = df_cleaned['Name'].isnull()
name_fill_values = [name_fill_list[i % len(name_fill_list)] for i in range(missing_name_indices.sum())]
df_cleaned.loc[missing_name_indices, 'Name'] = name_fill_values
```

```python
# Fill the missing values in the 'Age' column with 21
df_cleaned['Age'] = df_cleaned['Age'].fillna(21)
```

```python
# Fill the missing score values with mean value
df_cleaned['Score'] = df_cleaned['Score'].fillna(df_cleaned['Score'].mean())

print("Cleaned dataset:")
print(df_cleaned)
```

```python
# save the file to csv
df_cleaned.to_csv('cleaned_data.csv', index=False)
```
