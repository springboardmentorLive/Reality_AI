```python
import pandas as pd

df = pd.read_csv("infosys_springboard_dataset.csv")
data = pd.DataFrame(df)
print(data)
```

```python
data_cleaned = data.drop_duplicates()
data_cleaned
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.bar(data_cleaned["Student"], data_cleaned["Hours"],
        color=['red', 'green', 'yellow', 'blue', 'black', 'orange'])

plt.xlabel("Student")
plt.ylabel("Hours Spent")
plt.title("Infosys Springboard Internship - Cleaned Dataset")

plt.show()
```
