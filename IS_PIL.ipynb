```python
# Importing the PIL(Python Imaging Library) Library
from PIL import Image
from PIL import ImageDraw

# Open an Image
img = Image.open('/content/cute bb.jpg')

# Call draw Method to add 2d graphics in an image
I1 = ImageDraw. Draw(img)

# Add text to an image
I1.text((10,16), "Baby", fill=(105, 100, 0))

# Display edited image
img.show()

# Save the edited image
img.save("baby.png")
```

```python
from PIL import Image

# Open an image
image = Image.open('/content/cute bb.jpg')

# Display the image
image.show()

# Resize the image
resized_image = image.resize((200, 200))

# Save the image
resized_image.save("resized_eg.jpg")
```

```python
rotated_image = image.rotate(145)
rotated_image
```

```python
cropped_image = image.crop((10, 10, 100, 100))  #(LEFT, TOP, RIGHT, BOTTOM)
cropped_image
```

```python
grayscale_image = image.convert("L")  # LUMINE : L is used to convert color image into grayscale image
grayscale_image
```
