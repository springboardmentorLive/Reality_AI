```python
# Importing Necessary Libraries
from skimage import data
from skimage.color import rgb2gray
import matplotlib.pyplot as plt

# Setting the plot size to 15, 15
plt.figure(figsize=(15,15))

# Sample Image of scikit-image package
coffee = data.astronaut()
plt.subplot(1, 2, 1)  # ROWS, COLUMNS, PLOT

# Displaying the sample image
plt.imshow(coffee)

# Converting RGB to image to Monochromne
gray_coffee = rgb2gray(coffee)
plt.subplot(1, 2, 2)

# Displaying the image in Format
plt.imshow(gray_coffee, cmap="gray")
```

```python
# Importing Necessary Libraries
from skimage import data
from skimage.color import rgb2hsv
import matplotlib.pyplot as plt

# Setting the plot size to 15, 15
plt.figure(figsize=(15,15))

# Sample Image of scikit-image package
coffee = data.cat()
plt.subplot(1, 2, 1)  # ROWS, COLUMNS, PLOT

# Displaying the sample image
plt.imshow(coffee)

# Converting RGB to image to Monochromne
hsv_coffee = rgb2hsv(coffee)
plt.subplot(1, 2, 2)

# Displaying the image in Format
hsv_coffee_colorbar = plt.imshow(hsv_coffee)

# pad = space between colorbar and image (more the value more the space)
# fraction = thickness of the color bar (more the value more the thickness)
plt.colorbar(hsv_coffee_colorbar, fraction = 0.06, pad=0.05)
```
