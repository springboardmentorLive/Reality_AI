```python
import pandas as pd

left = pd.DataFrame(
    {
        "key":["k0","k1","k2","k3"],
        "A":["A0","A1","A2","A3"],
        "B":["B0","B1","B2","B3"],
    }
)

right = pd.DataFrame(
    {
        "key":["k0","k1","k2","k3"],
        "C":["C0","C1","C2","C3"],
        "D":["D0","D1","D2","D3"],
    })

res = pd.merge(left,right,on="key")
print(res)
```

```python
left = pd.DataFrame(
    {
        "key1":["k0","k1","k2","k3"],
        "key2":["k0","k1","k2","k3"],
        "A":["A0","A1","A2","A3"],
        "B":["B0","B1","B2","B3"],
    }
)

right = pd.DataFrame(
    {
        "key1":["k0","k1","k2","k3"],
        "key2":["k0","k1","k2","k3"],
        "C":["C0","C1","C2","C3"],
        "D":["D0","D1","D2","D3"],
    })
```

```python
res = pd.merge(left,right,how="left",on=["key1","key2"])
print(res)
```

```python
res = pd.merge(left,right,how="right",on=["key1","key2"])
print(res)
```

```python
res = pd.merge(right,left,how="outer",on=["key1","key2"])
print(res)
```

```python
res = pd.merge(left,right,how="inner",on=["key1","key2"])
print(res)
```

## Day 4: 18/11

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv(r"/content/User_Data.csv")
dataset.head()
```

```python
x = dataset.iloc[:,[2,3]].values
y = dataset.iloc[:,4].values

x
# y
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.25,random_state=42,shuffle=True)
```

```python
from sklearn.preprocessing import StandardScaler
sc_x = StandardScaler()
xtrain = sc_x.fit_transform(x_train)
xtest = sc_x.transform(x_test)

print(xtrain[0:10,:])
```

```python
from sklearn.linear_model import LogisticRegression

classifier = LogisticRegression(random_state=0)
classifier.fit(xtrain,y_train)
```

```python
y_pred = classifier.predict(xtest)
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test,y_pred)

print("Confusion Matrix:\n", cm)
```

```python
from sklearn.metrics import accuracy_score
print("Accuracy: ",accuracy_score(y_test,y_pred))
```

```python
from matplotlib.colors import ListedColormap

X_set, y_set = xtest, y_test
X1, X2 = np.meshgrid(np.arange(start = X_set[:, 0].min() - 1,
                               stop = X_set[:, 0].max() + 1, step = 0.01),
                     np.arange(start = X_set[:, 1].min() - 1,
                               stop = X_set[:, 1].max() + 1, step = 0.01))

plt.contourf(X1, X2, classifier.predict(
             np.array([X1.ravel(), X2.ravel()]).T).reshape(
             X1.shape), alpha = 0.2, cmap = ListedColormap(('red', 'green'))) # alpha sem transparent

plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())

for i, j in enumerate(np.unique(y_set)):
    plt.scatter(X_set[y_set == j, 0], X_set[y_set == j, 1],
                c = ListedColormap(('red', 'green'))(i), label = j)

plt.title('Classifier (Test set)')
plt.xlabel('Age')
plt.ylabel('Estimated Salary')
plt.legend()
plt.show()
```

```python
import cv2
import os

#Path to your video file
video_path="/content/seal-dance.gif"

#Folder to save extracted frames
output_folder = "frames1"

#Create the folder if it doesn't exist
os.makedirs(output_folder, exist_ok=True)

#Load the video
cap = cv2.VideoCapture("/content/seal-dance.gif")

#Check if video opened successfully
if not cap.isOpened():
    print("Error: Could not open video. Please ensure the video path is correct and the file exists.")
    video_opened = False
else:
    video_opened = True

if video_opened:
    frame_count=0
    max_frames=10
    while True:
        ret, frame1=cap.read()
        if not ret:
            break #End of video
        if frame_count>=max_frames:
          break
        #Save frame as image
        frame_filename = os.path.join(output_folder, f"frame1_{frame_count:03d}.jpg")
        cv2.imwrite(frame_filename, frame1)
        print(f"Saved {frame_filename}")
        frame_count += 1

    #Release resources
    cap.release()
    cv2.destroyAllWindows()
```
