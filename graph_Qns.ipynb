```python
import pandas as pd
import matplotlib.pyplot as plt
data = {
    "Name": ["Alice", "Bob", "Charlie", "Bob", "Alice"],
    "Age": [25, 30, 35, 30, 25],
    "Salary": [50000, 60000, 70000, 60000, 50000]
}
df = pd.DataFrame(data)
print("Original Dataset:")
print(df)
```

```python
df_clean = df.drop_duplicates()

print("\nDataset After Removing Duplicates:")
print(df_clean)


```

```python

df_clean.plot(
    x="Name",
    y="Salary",
    kind="bar",
    figsize=(6,4),
    legend=False
)

plt.title("Salary of Employee (Duplicates Removed)")
plt.xlabel("Name")
plt.ylabel("Salary")
plt.show()

```
