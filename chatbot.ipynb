```python
print("Chatbot: Hi! I am your chatbot. Type 'bye' to exit.")

while True:
    user = input("You: ").lower().strip()

    if user in ["hi", "hello", "hey"]:
        print("Chatbot: Hello! How can I help you?")

    elif user in ["how are you", "how r u", "how are u"]:
        print("Chatbot: I am fine! Thank you 😊")

    elif user == "bye":
        print("Chatbot: Goodbye! Have a nice day! 👋")
        break

    else:
        print("Chatbot: Sorry, I don't understand.")
```

```python
print("Chatbot: Hi! I'm your AI-style chatbot. Type 'bye' to exit.\n")

intents = {
    "greeting": ["hi", "hello", "hey"],
    "how_are_you": ["how are you", "how r u", "how are u"],
    "name": ["your name", "who are you", "what is your name"],
    "thanks": ["thank", "thanks", "thnx"],
    "weather": ["weather", "temperature"],
    "age": ["age", "old"]
}

responses = {
    "greeting": "Hello! How can I assist you today?",
    "how_are_you": "I'm doing great! Thanks for asking 😊",
    "name": "I'm a Python chatbot created by you!",
    "thanks": "You're welcome!",
    "weather": "Can't check weather yet, but I hope it's sunny ☀️",
    "age": "I don't age… I'm just code 😄"
}

while True:
    user = input("You: ").lower().strip()

    if user == "bye":
        print("Chatbot: Goodbye! 👋")
        break

    found = False
    for intent, keywords in intents.items():
        for word in keywords:
            if word in user:   # keyword present?
                print("Chatbot:", responses[intent])
                found = True
                break
        if found:
            break

    if not found:
        print("Chatbot: Sorry, I didn't get that.")
```

```python
class ChatBot:
    def __init__(self):
        # Map: answer -> list of possible user questions
        self.responses = {
            "My name is ChatBot!": [
                "what is your name",
                "what's your name",
                "who are you",
                "tell me your name",
                "your name"
            ],

            "I'm doing great, thanks!": [
                "how are you",
                "how are you doing",
                "how do you feel",
                "are you ok",
                "how's it going"
            ],

            "I'm a chatbot created to help you!": [
                "what are you",
                "what do you do",
                "what is your purpose",
                "who created you",
                "why are you here"
            ],

            "I can help you with anything!": [
                "help me",
                "can you help me",
                "i need help",
                "assist me"
            ],

            "Goodbye! Have a nice day!": [
                "bye",
                "goodbye",
                "see you",
                "tata"
            ]
        }

    def get_response(self, user_input):
        user_input = user_input.lower().strip()

        # Check each answer group
        for answer, questions in self.responses.items():
            for q in questions:
                if q in user_input:
                    return answer

        # Default response
        return "Sorry, I didn't understand that. Can you rephrase?"

# ------------------ RUN THE CHATBOT ------------------

bot = ChatBot()

print("ChatBot: Hello! Ask me anything.")

while True:
    user_input = input("You: ")

    if user_input.lower() in ["bye", "exit", "quit"]:
        print("ChatBot: Goodbye! Have a nice day!")
        break

    response = bot.get_response(user_input)
    print("ChatBot:", response)
```
