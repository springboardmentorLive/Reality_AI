Area Plot (series.plot.area): This plot displays the magnitude of values over time or ordered categories. It's similar to a line plot, but the area between the line and the axis is filled in, making it useful for showing accumulation or the proportion of total over time.

Horizontal Bar Plot (series.plot.barh): This plot presents categorical data with rectangular bars, where the length of each bar is proportional to the value it represents. It's useful for comparing values across different categories, especially when category names are long or there are many categories.

Box Plot (series.plot.box): Also known as a box-and-whisker plot, it displays the distribution of numerical data and skewness through five key summary numbers: the minimum, first quartile (Q1), median (Q2), third quartile (Q3), and maximum. It's excellent for identifying outliers and understanding the spread of data.

Histogram (series.plot.hist): A histogram represents the distribution of a numerical dataset. It groups numbers into ranges (bins) and plots bars showing how many values fall into each range. It helps you see the shape, spread, and central tendency of your data.

KDE Plot (series.plot.kde) / Density Plot (series.plot.density): These are essentially the same (density is an alias for kde in pandas plotting). A Kernel Density Estimate (KDE) plot visualizes the probability density function of a continuous random variable. It's a smoothed histogram, providing a continuous curve that shows where values are more concentrated.

Pie Plot (series.plot.pie) **bold text**: This circular statistical graphic is divided into slices to illustrate numerical proportion. In your case, each slice represents a proportion of the total sum of the series values. It's best used for showing parts of a whole.

Line Plot (series.plot.line): This plot displays information as a series of data points called 'markers' connected by straight line segments. It's commonly used to visualize trends over time or ordered categories, showing how one variable changes in relation to another.

# Task
Create a pandas DataFrame, intentionally add duplicate rows to it, then identify and remove these duplicate rows, and finally display the original DataFrame, the identified duplicates, and the DataFrame after duplicate removal.

## Create Dataset Which have  Duplicates Rows.

I create 5x5 pandas DataFrame firstly add duplicate rows.



```python
import pandas as pd
import numpy as np

# 1. Create a dictionary that can be converted into a pandas DataFrame (5x5)
data = {
    'Age': [19, 22, 30, 19, 22],
    'Name': ['Akshat', 'Bhavika', 'Chandra', 'Akshat', 'Bhavika'],
    'Gender': ['M', 'F', 'M', 'M', 'F']
}

df_original = pd.DataFrame(data)
duplicate_rows = df_original.iloc[[0, 2]].copy()
df_original = pd.concat([df_original, duplicate_rows], ignore_index=True)
print("DataFrame with intentional duplicates (df_original):")
print(df_original)
```

```python
import pandas as pd

# 1. Use .duplicated(keep=False) to create a boolean Series that marks all occurrences of duplicate rows as True
duplicate_mask = df_original.duplicated(keep=False)

# 2. Filter df_original using the duplicate_mask to extract only the rows that are duplicates
df_duplicates = df_original[duplicate_mask]

# 3. Print df_duplicates to display the identified duplicate rows
print("Identified duplicate rows (df_duplicates):")
print(df_duplicates)
```

## Remove Duplicate Rows

```python
import pandas as pd

# 1. Use drop_duplicates() to remove duplicate rows, keeping the first occurrence
df_no_duplicates = df_original.drop_duplicates(keep='first')

# 2. Print the df_no_duplicates DataFrame
print("DataFrame after duplicate removal (df_no_duplicates):")
print(df_no_duplicates)
```

```python
print("Original DataFrame with duplicates (df_original):")
print(df_original)

print("\nIdentified duplicate rows (df_duplicates):")
print(df_duplicates)

print("\nDataFrame after duplicate removal (df_no_duplicates):")
print(df_no_duplicates)
```
