```python
import pandas as pd

left = pd.DataFrame(
    {
        "key":["k0","k1","k2","k3"],
        "A":["A0","A1","A2","A3"],
        "B":["B0","B1","B2","B3"],
    }
)

right = pd.DataFrame(
    {
        "key":["k0","k1","k2","k3"],
        "C":["C0","C1","C2","C3"],
        "D":["D0","D1","D2","D3"],
    })

res = pd.merge(left,right,on="key")
print(res)
```

```python
left = pd.DataFrame(
    {
        "key1":["k0","k1","k2","k3"],
        "key2":["k0","k1","k2","k3"],
        "A":["A0","A1","A2","A3"],
        "B":["B0","B1","B2","B3"],
    }
)

right = pd.DataFrame(
    {
        "key1":["k0","k1","k2","k3"],
        "key2":["k0","k1","k2","k3"],
        "C":["C0","C1","C2","C3"],
        "D":["D0","D1","D2","D3"],
    })
```

```python
res = pd.merge(left,right,how="left",on=["key1","key2"])
print(res)
```

```python
res = pd.merge(left,right,how="right",on=["key1","key2"])
print(res)
```

```python
res = pd.merge(right,left,how="outer",on=["key1","key2"])
print(res)
```

```python
res = pd.merge(left,right,how="inner",on=["key1","key2"])
print(res)
```
