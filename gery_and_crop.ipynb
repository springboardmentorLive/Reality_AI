```python
from PIL import Image

#Open an image
image = Image.open("flower.jpg")

#Display the image
image.show()

#Resize the image
resized_image =  image.resize((200, 200))

#Save the image
resized_image.save("resized example.jpg")
```

```python
rotated_image = image.rotate(45)

rotated_image
```

```python
cropped_image = image.crop((10, 10, 100, 100)) # left,top,right,bottom
cropped_image
```

```python
grayscale_image = image.convert("L")#lumine
grayscale_image
```
